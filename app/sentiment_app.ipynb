```python
import streamlit as st
from transformers import DistilBertTokenizerFast, DistilBertForSequenceClassification
import torch
```

```python
# Load model and tokenizer
@st.cache_resource
def load_model():
    model = DistilBertForSequenceClassification.from_pretrained("../models/distilbert-sentiment")
    tokenizer = DistilBertTokenizerFast.from_pretrained("../models/distilbert-sentiment")
    return model, tokenizer

model, tokenizer = load_model()
```

```python
# Label Map
label_map = {0: "Negative", 1: "Neutral", 2: "Positive"}

# App Title
st.title("✈️ Airline Sentiment Classifier")
st.write("Enter a tweet about an airline, and the model will predict its sentiment.")
```

```python
# Text Input
user_input = st.text_area("Enter tweet text:")

if st.button("Analyze Sentiment"):
    if user_input:
        inputs = tokenizer(user_input, return_tensors="pt", truncation=True, padding=True)
        with torch.no_grad():
            outputs = model(**inputs)
            pred = torch.argmax(outputs.logits).item()
            st.success(f"Predicted Sentiment: **{label_map[pred]}**")
    else:
        st.warning("Please enter some text.")
```
