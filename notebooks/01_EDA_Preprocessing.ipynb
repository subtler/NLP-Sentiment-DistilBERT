```python
# 📦 Step 1: Import required libraries
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
# Load the dataset
df= pd.read_csv('../data/Tweets.csv')
df = df[['text','airline_sentiment']]# we only need these two
df
```

```python
# Check class distribution
print(df['airline_sentiment'].value_counts())
```

```python
#plot class distribution
sns.countplot(df, x='airline_sentiment')
plt.title('Class distribution')
plt.show()
```

```python
#Encode labels
# Why? ML models don't understand "positive", "negative" — they need numbers
label_map={'negative':0,'neutral':1,'positive':2}
df['label']=df['airline_sentiment'].map(label_map)
df.head()
```

```python
df.dropna(inplace=True)
```

```python
df[['text','label']].to_csv('../data/clean_tweets',index=False)
print("✅ Cleaned data saved to data/clean_tweets.csv")
```
